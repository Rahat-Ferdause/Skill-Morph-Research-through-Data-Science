```python
drive.mount('/content/drive')
gym = pd.read_csv('/content/drive/MyDrive/Datasets/gym_membership.csv')  # No header in raw CSV
print("Dataset loaded successfully. Shape:", gym.shape)
```

```python
  # Upload the Gym data in google drive

# Display last 5 rows

# Check data types
# Check data
print(gym.tail())
print(gym.shape)  # How many rows and columns
print(gym.columns.tolist())  # Column names
```

```python
# Task 1: Basic Matplotlib (3 plots)
# 1. Bar chart of gym membership types
# Use: gym['abonoment_type'].value_counts()
# Count abonoment_type
abonoment_count = gym['abonoment_type'].value_counts()
# Simple bar chart
# plt.bar(['Standard', 'Premium'], abonoment_count.values)
# plt.title('Abonoment Count')
# plt.show()

    # use the index from value_counts() to automatically get labels
plt.bar(abonoment_count.index, abonoment_count.values, color=['blue', 'red'])
plt.title('Abonoment Count')
plt.xlabel('Abonoment')
plt.ylabel('Number of Abonoment')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=1)
plt.savefig("/content/drive/MyDrive/Datasets/Abonoment Count.png", dpi=300, bbox_inches='tight')  # Save
# Add value labels on top of bars
for i, v in enumerate(abonoment_count.values):
    plt.text(i, v + 0.5, str(v), ha='center', va='bottom', fontweight='bold')
# Save the chart
plt.savefig("/content/drive/MyDrive/Datasets/Abonoment Count.png", dpi=300, bbox_inches='tight')
plt.show()



# 2. Histogram of time in gym
# Use: plt.hist(gym['avg_time_in_gym'])
plt.hist(gym['avg_time_in_gym'])
plt.title('Average Time in Gym')
plt.xlabel('Time in Gym')
plt.ylabel('Frequency')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=1)
plt.savefig("/content/drive/MyDrive/Datasets/Average Time in Gym.png", dpi=300, bbox_inches='tight')
plt.show()


# 3. Scatter plot: Age vs visits per week
# Use: plt.scatter(gym['Age'], gym['visit_per_week'])
plt.scatter(gym['Age'], gym['visit_per_week'])
plt.title('Age vs Visits per Week')
plt.xlabel('Age')
plt.ylabel('Visits per Week')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=1)
plt.savefig("/content/drive/MyDrive/Datasets/Age vs Visits per Week.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 1. Box plot: Gender vs Age
# Use: sns.boxplot(data=gym, x='gender', y='Age')
#plt.figure(figsize=(8, 6))
sns.boxplot(data=gym, x='gender', y='Age')
plt.title('Gender vs Age')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.savefig("/content/drive/MyDrive/Datasets/Gender vs Age.png", dpi=300, bbox_inches='tight')
plt.show()

# 2. Count plot: Days per week
# Use: sns.countplot(data=gym, x='days_per_week')
sns.countplot(data=gym, x='days_per_week')
plt.title('Days per Week')
plt.xlabel('Days per Week')
plt.ylabel('Count')
plt.xticks(rotation=90)
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.savefig("/content/drive/MyDrive/Datasets/Days per Week.png", dpi=300, bbox_inches='tight')
plt.show()

# 3. Bar plot: Average visits by membership type
# Use: sns.barplot(data=gym, x='abonoment_type', y='visit_per_week')
sns.barplot(data=gym, x='abonoment_type', y='visit_per_week')
plt.title('Average Visits by Membership Type')
plt.xlabel('Membership Type')
plt.ylabel('Average Visits')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.savefig("/content/drive/MyDrive/Datasets/Average Visits by Membership Type.png", dpi=300, bbox_inches='tight')
plt.show()
```
